```python
import os
import pandas as pd

# 1. Crear una copia de los datos originales
df_simple = df.copy()

# 2. Convertir 'mmr' y 'sellingprice' a numérico y eliminar nulos
df_simple['mmr'] = pd.to_numeric(df_simple['mmr'], errors='coerce')
df_simple['sellingprice'] = pd.to_numeric(df_simple['sellingprice'], errors='coerce')
df_simple = df_simple.dropna(subset=['sellingprice', 'mmr'])

# 3. Limpiar estados a mayúsculas y filtrar códigos inconsistentes (que no tengan 2 letras)
df_simple['state'] = df_simple['state'].astype(str).str.upper().str.strip()

df_simple = df_simple[df_simple['state'].str.len() == 2]

# 4. Limpiar fechas (reemplazando la columna 'saledate' por una limpia de formato YYYY-MM-DD)
def parse_saledate_simple(date_str):
    try:
        parts = str(date_str).split()
        if len(parts) >= 4:
            date_part = f"{parts[1]} {parts[2]} {parts[3]}"
            return pd.to_datetime(date_part, format='%b %d %Y').strftime('%Y-%m-%d')
    except:
        return None

df_simple['saledate'] = df_simple['saledate'].apply(parse_saledate_simple)
df_simple = df_simple.dropna(subset=['saledate'])

# 5. Guardar el archivo limpio sin columnas adicionales
simple_output_path = '/content/car_prices_limpio.csv'
df_simple.to_csv(simple_output_path, index=False)

# 6. Validar tamaño final del archivo
tamano_bytes = os.path.getsize(simple_output_path)
tamano_mb = tamano_bytes / (1024 * 1024)

print(f"¡Limpieza sin campos adicionales completada!")
print(f"Filas resultantes: {len(df_simple)}")
print(f"Tamaño del nuevo archivo: {tamano_mb:.2f} MB")


print(f"¡Limpieza completada con éxito! Archivo guardado en: {simple_output_path}")
print(f"Filas originales: {len(df)} | Filas limpias finales: {len(df_simple)}")
```

```python
# Vamos a identificar exactamente cuáles fueron los 38 registros eliminados y la razón

# 1. Filas originales
filas_originales = set(df.index)

# 2. Filas que sobrevivieron a la limpieza
filas_limpias = set(df_simple.index)

# 3. Filas eliminadas
filas_eliminadas_indices = filas_originales - filas_limpias
df_eliminados = df.loc[list(filas_eliminadas_indices)].copy()

print(f"Total de registros eliminados: {len(df_eliminados)}\n")

# 4. Clasificar por qué se eliminaron
nulos_precio_o_mmr = df_eliminados[df_eliminados['sellingprice'].isnull() | df_eliminados['mmr'].isnull()]

# Intentar parsear la fecha en los eliminados para ver si falló por fecha
def test_date(val):
    try:
        parts = str(val).split()
        if len(parts) >= 4:
            pd.to_datetime(f"{parts[1]} {parts[2]} {parts[3]}", format='%b %d %Y')
            return False
    except:
        pass
    return True

nulos_fecha = df_eliminados[df_eliminados['saledate'].apply(test_date)]
estado_incorrecto = df_eliminados[df_eliminados['state'].astype(str).str.strip().str.len() != 2]

print(f"- Eliminados por precio o MMR vacío (nulo): {len(nulos_precio_o_mmr)}")
print(f"- Eliminados por fecha con formato corrupto o vacía: {len(nulos_fecha)}")
print(f"- Eliminados por código de estado incorrecto (diferente de 2 letras): {len(estado_incorrecto)}")

# Mostrar una muestra de los registros eliminados para comprobarlos
if len(df_eliminados) > 0:
    print("\nMuestra de las filas eliminadas:")
    display(df_eliminados[['state', 'saledate', 'sellingprice', 'mmr']].head(10))
```

### **Conclusión del Proceso de Limpieza, Saneamiento y Preparación de Datos (Versión Final)**

Para asegurar la máxima calidad de los datos y un rendimiento óptimo en la integración con **Google Looker Studio**, se implementó un pipeline de preparación estructurado en dos fases consecutivas:

---

#### **Fase 1: Depuración de Registros Críticos e Inconsistencias Estructurales**
1. **Lectura e Inspección Inicial:** Se importó el conjunto de datos original con **558,837 registros**, identificando columnas con datos nulos y discrepancias de formato.
2. **Saneamiento Financiero (`sellingprice` y `mmr`):** Se forzó la conversión a tipo numérico de los precios de venta y la tasación MMR, lo que permitió identificar y remover únicamente **38 registros** que carecían de estas métricas esenciales para la construcción de KPIs económicos.
3. **Normalización Geográfica (`state`):** Se transformaron los códigos de estado a mayúsculas (`str.upper()`) y se descartaron los registros cuyas abreviaturas no correspondían al estándar oficial de dos caracteres.
4. **Conversión Temporal (`saledate`):** Se transformaron las complejas cadenas de fecha originales (con husos horarios extensos) a un formato de fecha simple estándar (`YYYY-MM-DD`), esencial para análisis evolutivos en series de tiempo.

---

#### **Fase 2: Tratamiento de Vacíos, Codificación Corrupta y Homogeneidad Categórica**
Para evitar la pérdida masiva de información útil que hubiese provocado una eliminación en cadena de registros, se aplicaron tratamientos de imputación y corrección de codificación sobre las variables secundarias:

1. **Imputación Estándar (`make`, `model`, `trim`, `body`):** Se completaron las celdas vacías utilizando la etiqueta **`"No Especificado"`**, preservando el valor comercial de las filas.
2. **Ajuste de Atributo de Uso (`odometer`):** Los registros faltantes en el odómetro se completaron asignándoles un valor de **`0`**, asegurando la integridad del campo numérico.
3. **Neutralización de Errores de Codificación (`color` e `interior`):** Se identificaron y reemplazaron tanto los nulos reales como más de **41,000 celdas** que contenían caracteres de codificación rotos (tales como `â€”`, `—` o `~`), estandarizándolos bajo el término **`"No Especificado"`**.
4. **Normalización de Vendedores (`seller`):** Se homogeneizaron los nombres de las agencias y vendedores llevándolos a un formato uniforme en mayúsculas y libre de espacios en blanco redundantes, conservando el nivel de detalle individual de cada actor comercial en el dataset.

---

**Resultado Final:** El archivo depurado **`car_prices_limpio_v2.csv`** consta de **558,799 registros**, libre de nulos en variables clave, corregido en sus problemas de formato de texto/fecha, y listo para ser explotado en la creación de tableros de control de alta fidelidad.
